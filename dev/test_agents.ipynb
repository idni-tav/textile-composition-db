```python
from litellm import completion

from transform.semantic.prompts import COMPOSITION_PROMPT
from transform.semantic.schemas import MaterialComposition


response = completion(
    model="ollama_chat/llama3.1:8b",
    messages=[
        {
            "role": "system",
            "content": COMPOSITION_PROMPT,
        },
        {
            "role": "user",
            "content": materials_raw,
        },
    ],
)

content = response.choices[0].message.content

print(content)

composition = MaterialComposition.model_validate_json(content)

print(composition)
```

```python
import json

from transform.semantic.agents.attribution_agent import build_attribution_agent
from transform.semantic.schemas import FinalMaterialProfile

# ------------------------------------------------------------------
# Test inputs
# ------------------------------------------------------------------

task = {
    "extracted": {
        "materials": [
            {
                "name": "Cotton",
                "percentage": 95
            },
            {
                "name": "Elastane",
                "percentage": 5
            }
        ],
        "certification_name": [
            "OEKO-TEX Standard 100"
        ],
        "origins": [
            "Portugal"
        ]
    },
    "materials_raw": """
Made from 95% Cotton, 5% Elastane.
OEKO-TEX Standard 100 certified.
Made in Portugal.
"""
}

# ------------------------------------------------------------------
# Build agent
# ------------------------------------------------------------------

attribution_agent = build_attribution_agent()

# ------------------------------------------------------------------
# Run agent
# ------------------------------------------------------------------

response = attribution_agent.run(
    json.dumps(task, indent=2)
)

# ------------------------------------------------------------------
# Inspect raw output
# ------------------------------------------------------------------

print("===== RAW RESPONSE =====")
print(type(response))
print(repr(response))
print(response)

# ------------------------------------------------------------------
# Validate output
# ------------------------------------------------------------------

profile = FinalMaterialProfile.model_validate(response)

print("\n===== VALIDATED OUTPUT =====")
print(profile)
```
