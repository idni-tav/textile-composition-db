```python

import ollama
import json



product_text = """
Made from a silky lycra that has beautiful drape & 4 way stretch.\nMade from a silky lycra that has beautiful drape & 4 way stretch.\n| 95% Polyester, 5% Lycra | Wash with similar colours, the dye may run.\n| 95% Polyester, 5% Lycra | Wash with similar colours, the dye may run.\nMade in London.\nMade in London.
"""



prompt = f"""
You are a semantic information extraction system for fashion products.

Extract ONLY information that is explicitly stated in the product text.

Return a JSON object with exactly these fields:

{{
    "composition": [
        {{
            "material": "string",
            "percentage": number
        }}
    ],
    "certifications": ["string"]
}}

Rules:

- Extract all explicitly stated material compositions and percentages.
- Do not invent or infer materials or percentages.
- Extract certifications only when they are explicitly mentioned.
- If no composition is explicitly stated, return an empty list.
- If no certifications are explicitly stated, return an empty list.
- Return ONLY valid JSON. Do not include explanations or markdown.

Product text:
---
{product_text}
---
"""


# --------------------------------------------------
# 3. Call local Ollama model
# --------------------------------------------------

response = ollama.chat(
    model="llama3.1:8b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
)


# --------------------------------------------------
# 4. Display raw response
# --------------------------------------------------

raw_output = response["message"]["content"]

print(raw_output)


# --------------------------------------------------
# 5. Try parsing the response as JSON
# --------------------------------------------------

try:
    extracted = json.loads(raw_output)

    print("\nParsed JSON:")
    print(json.dumps(extracted, indent=2))

except json.JSONDecodeError as e:
    print("\nThe model did not return valid JSON.")
    print(e)

```
